## 1. ExactMemory – Cryptographic, Tamper‑Evident

```python
from recallspection import ExactMemory

exact = ExactMemory()
exact.add("user_123", {"theme": "dark", "language": "en"})
print("✅ Stored:", exact.get("user_123"))

# Tamper test
exact._storage["user_123"] = b"TAMPERED"
result = exact.get("user_123")
print("🔒 Tamper detection:", result)  # Should print None
```

## 2. SWSTM – Neural Memory with Fuzzy Retrieval

```python
from recallspection import SWSTMEngine

swstm = SWSTMEngine(mode="flat", flat_num_slots=200)

swstm.add("capital of France", "Paris")
swstm.add("capital of Germany", "Berlin")

print("Exact:", swstm.get("capital of France"))
print("Paraphrase:", swstm.get("France's capital"))
print("Paraphrase:", swstm.get("German capital"))
```
