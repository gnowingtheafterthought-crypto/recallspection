## 1. ExactMemory – Cryptographic, Tamper‑Evident

```python
from recallspection import ExactMemory

# Create a tamper‑proof memory
exact = ExactMemory()

# Store a fact
exact.add("user_123", {"theme": "dark", "language": "en"})
print("✅ Stored:", exact.get("user_123"))

# Tamper test – manually corrupt the value
exact._storage["user_123"] = b"TAMPERED"
result = exact.get("user_123")
print("🔒 Tamper detection:", result)  # Should print None
```

## 2. SWSTM – Neural Memory with Fuzzy Retrieval

```python
from recallspection import SWSTMEngine

# Create a neural memory (auto‑mode: flat → hierarchical → PQ)
swstm = SWSTMEngine(mode="flat", flat_num_slots=200)

# Add facts
swstm.add("capital of France", "Paris")
swstm.add("capital of Germany", "Berlin")
swstm.add("capital of Italy", "Rome")

# Retrieve with exact match
print("Exact:", swstm.get("capital of France"))  # ['Paris']

# Retrieve with paraphrase (fuzzy query)
print("Paraphrase:", swstm.get("France's capital"))   # ['Paris']
print("Paraphrase:", swstm.get("German capital"))     # ['Berlin']
print("Paraphrase:", swstm.get("What is the capital of Italy?"))  # ['Rome']
```
